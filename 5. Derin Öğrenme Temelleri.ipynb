# **Evrişimli Sinir Ağları (CNN) Nedir?**

CNN'ler, bilgisayarlara "görmeyi" öğreten mimarilerdir. Fotoğrafı dümdüz etmek yerine, üzerinde **"Sihirli Büyüteçler" (Filtreler/Kerneller)** gezdirerek resmin içindeki desenleri ararlar.



### CNN'in 3 Temel Silahı:

1. **Evrişim (Convolution) Katmanı:** Matematiksel bir büyüteçtir (genellikle 3x3 piksel boyutunda). Resmin üzerinde piksel piksel kayarak dolaşır. Amacı resmi anlamak değil, resimdeki **düz çizgileri, köşeleri, renk geçişlerini** (kenar tespiti) bulmaktır.
   Matematiği iki matrisin çarpımıdır (Nokta çarpım):
   $$S(i, j) = (I * K)(i, j) = \sum_m \sum_n I(i+m, j+n) K(m, n)$$

2. **Ortaklama / Havuzlama (Max Pooling) Katmanı:**
   Çözünürlük düşürme (Sıkıştırma) işlemidir. 2x2'lik bir alana bakar ve sadece "en parlak/en önemli" pikseli alıp diğerlerini çöpe atar. Bu sayede resim küçülür, gereksiz detaylar (gürültü) silinir ve bilgisayarın hızı muazzam artar.

3. **Tam Bağlı (Fully Connected / Dense) Katman:**
   Filtreler "Burada dik bir çizgi ve yuvarlak bir cisim var" dedikten sonra, bu bulguları birleştirip "O zaman bu bir KEDİDİR!" kararını veren, klasik sinir ağı katmanıdır (Karar mercii).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d
from skimage import data, color

print("="*75)
print("1. AŞAMA: YAPAY ZEKA GÖZÜYLE DÜNYAYA BAKMAK (MANUEL EVRİŞİM)")
print("="*75)

# 1. Örnek bir fotoğraf yüklüyoruz ve siyah-beyaza çeviriyoruz (Bilgisayar tek kanalı daha kolay anlar)
orijinal_resim = color.rgb2gray(data.astronaut())

# 2. YAPAY ZEKANIN BÜYÜTEÇLERİ (Filtreler/Kerneller)
# Bu 3x3 matrisler tamamen matematiksel sihirbazlardır!

# Dikey Kenar Bulucu (Sobel Y): Sadece dikey çizgileri parlatır!
dikey_filtre = np.array([[-1, 0, 1], 
                         [-2, 0, 2], 
                         [-1, 0, 1]])

# Yatay Kenar Bulucu (Sobel X): Sadece yatay çizgileri parlatır!
yatay_filtre = np.array([[-1, -2, -1], 
                         [ 0,  0,  0], 
                         [ 1,  2,  1]])

# 3. EVRİŞİM (CONVOLUTION) İŞLEMİNİ UYGULAMA
# Filtreleri resmin üzerinde gezdiriyoruz (Matris çarpımı)
dikey_ozellik_haritasi = convolve2d(orijinal_resim, dikey_filtre, mode='valid')
yatay_ozellik_haritasi = convolve2d(orijinal_resim, yatay_filtre, mode='valid')

# 4. GÖRSELLEŞTİRME
fig, eksenler = plt.subplots(1, 3, figsize=(18, 6))

eksenler[0].imshow(orijinal_resim, cmap='gray')
eksenler[0].set_title("Orijinal Resim (Bizim Gördüğümüz)", fontsize=14)

eksenler[1].imshow(dikey_ozellik_haritasi, cmap='gray')
eksenler[1].set_title("Dikey Filtre Çıktısı (Sadece Dikey Kenarlar!)", fontsize=14)

eksenler[2].imshow(yatay_ozellik_haritasi, cmap='gray')
eksenler[2].set_title("Yatay Filtre Çıktısı (Sadece Yatay Kenarlar!)", fontsize=14)

for ax in eksenler:
    ax.axis('off') # Eksenleri gizle
plt.show()

print("ANALİZ:")
print("Yapay zeka resme bütünüyle bakmaz. İlk katmanlarda bu şekilde sadece kenarları,")
print("kıvrımları ve dokuları bulur. Sonraki katmanlarda bu kenarları birleştirip")
print("'Göz', 'Kulak' gibi şekiller oluşturur!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d
from skimage import color, io  # 'io' modülünü ekledik (Input/Output)

print("="*75)
print("1. AŞAMA: KENDİ FOTOĞRAFINIZLA MANUEL EVRİŞİM")
print("="*75)

# ==========================================
# 1. KENDİ FOTOĞRAFINIZI YÜKLEME KISMI
# ==========================================
# DİKKAT: Fotoğrafınızı Jupyter Notebook'un çalıştığı klasöre sürükleyip bırakın.
# Sonra fotoğrafın tam adını (uzantısıyla beraber) aşağıya yazın.
fotograf_yolu = "resim.png"  # Örnek: "kedi.png", "vesikalik.jpg"

try:
    # Resmi dosyadan okuyoruz
    okunan_resim = io.imread(fotograf_yolu)
    
    # PNG gibi saydamlık (Alpha) kanalı olan 4 kanallı resimleri 3 kanala (RGB) düşürüyoruz
    if len(okunan_resim.shape) == 3 and okunan_resim.shape[-1] == 4:
        okunan_resim = okunan_resim[:, :, :3]
        
    # Resmi siyah-beyaza çeviriyoruz (Büyüteçlerimiz tek kanalda çalışır)
    orijinal_resim = color.rgb2gray(okunan_resim)

except FileNotFoundError:
    print(f"HATA: '{fotograf_yolu}' adında bir dosya bulamadım!")
    print("Lütfen fotoğrafı Jupyter klasörünüze yüklediğinizden ve adını yukarıya doğru yazdığınızdan emin olun.")
    raise # Kodu burada durdur

# 2. YAPAY ZEKANIN BÜYÜTEÇLERİ (Filtreler/Kerneller)
# Dikey Kenar Bulucu (Sobel Y)
dikey_filtre = np.array([[-1, 0, 1], 
                         [-2, 0, 2], 
                         [-1, 0, 1]])

# Yatay Kenar Bulucu (Sobel X)
yatay_filtre = np.array([[-1, -2, -1], 
                         [ 0,  0,  0], 
                         [ 1,  2,  1]])

# 3. EVRİŞİM (CONVOLUTION) İŞLEMİNİ UYGULAMA
dikey_ozellik_haritasi = convolve2d(orijinal_resim, dikey_filtre, mode='valid')
yatay_ozellik_haritasi = convolve2d(orijinal_resim, yatay_filtre, mode='valid')

# 4. GÖRSELLEŞTİRME
fig, eksenler = plt.subplots(1, 3, figsize=(18, 6))

eksenler[0].imshow(orijinal_resim, cmap='gray')
eksenler[0].set_title("Orijinal Resim (Sizin Seçiminiz)", fontsize=14, fontweight='bold')

eksenler[1].imshow(dikey_ozellik_haritasi, cmap='gray')
eksenler[1].set_title("Dikey Filtre (Sadece Dikey Kenarlar)", fontsize=14, fontweight='bold')

eksenler[2].imshow(yatay_ozellik_haritasi, cmap='gray')
eksenler[2].set_title("Yatay Filtre (Sadece Yatay Kenarlar)", fontsize=14, fontweight='bold')

for ax in eksenler:
    ax.axis('off')

plt.tight_layout()
plt.show()

print("\nANALİZ: Kendi fotoğrafınızdaki o pürüzsüz geçişlerin arka planda nasıl keskin")
print("matematiksel hatlara dönüştüğünü görebilirsiniz. Yapay zeka dünyayı işte böyle görür!")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np

print("\n" + "="*75)
print("2. AŞAMA: TENSORFLOW İLE CNN (EL YAZISI RAKAM TANIMA)")
print("="*75)

# 1. VERİ SETİNİ YÜKLEME VE HAZIRLAMA (70.000 adet el yazısı rakam)
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Veriyi 0-1 arasına ölçeklendiriyoruz (Sinir ağları küçük sayıları sever)
X_train = X_train / 255.0
X_test = X_test / 255.0

# CNN'ler resmin (Genişlik, Yükseklik, Renk Kanalı) formatında olmasını ister. 
# Siyah-beyaz olduğu için kanal = 1 (Renk olsaydı RGB için 3 olurdu)
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

# 2. CNN MİMARİSİNİ İNŞA ETME (Yapay Beyin Tasarımı)
model = models.Sequential([
    # ÖZELLİK ÇIKARMA BÖLÜMÜ (Dedektifler)
    # 32 tane 3x3 boyutunda büyüteç (filtre) gezdiriyoruz
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    
    # Resmi küçült ve en önemli bilgileri al (2x2'lik alanda en büyük değeri seç)
    layers.MaxPooling2D((2, 2)),
    
    # İkinci bir dedektif ekibi: 64 tane filtre (Detayları daha da derinlemesine inceler)
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # SINIFLANDIRMA BÖLÜMÜ (Karar Mercii)
    layers.Flatten(), # Resmi 2 boyutlu matristen tek boyutlu uzun bir çizgiye çeviriyoruz
    layers.Dense(64, activation='relu'), # Klasik Sinir Ağı
    layers.Dense(10, activation='softmax') # 0'dan 9'a kadar 10 farklı ihtimali (olasılığı) hesapla
])

# 3. MODELİ DERLEME VE EĞİTME
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("Yapay Zeka Okula Başladı (Eğitim Başlıyor)...")
# Sadece 3 döngü (epoch) eğitiyoruz ki sizi bekletmesin. Normalde bu 10-50 arası olur.
history = model.fit(X_train, y_train, epochs=3, validation_split=0.1)

print("\nEğitim Bitti! Test Setinde Başarı Oranını Ölçüyoruz:")
test_loss, test_acc = model.evaluate(X_test,  y_test, verbose=2)
print(f"CNN TEST BAŞARISI: %{test_acc*100:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("="*75)
print("4. AŞAMA: RASTGELE 10'LU GRUP (BATCH) TESTİ")
print("="*75)

# 1. Test setinden rastgele 10 farklı resim seçelim
num_images = 10
rastgele_indeksler = np.random.choice(len(X_test), num_images, replace=False)

# 2. Çizim için tablo hazırlığı (2 satır, 5 sütun)
plt.figure(figsize=(15, 7))

print("Model 10 rastgele resmi inceliyor...\n")

for i, idx in enumerate(rastgele_indeksler):
    resim = X_test[idx]
    gercek_deger = y_test[idx]
    
    # Modelden tahmin alıyoruz (verbose=0 ekranda gereksiz yazı çıkmasını engeller)
    tahmin_olasiliklari = model.predict(np.array([resim]), verbose=0)[0]
    tahmin_edilen = np.argmax(tahmin_olasiliklari)
    
    # 3. Sonuçları Ekrana Çizdirme
    plt.subplot(2, 5, i+1)
    plt.imshow(resim.reshape(28, 28), cmap='gray')
    
    # Karara göre renk ve ikon belirleme
    if tahmin_edilen == gercek_deger:
        renk = 'green'
        baslik = f"Yapay Zeka: {tahmin_edilen} \nGerçek: {gercek_deger}"
    else:
        renk = 'red'
        baslik = f"Yapay Zeka: {tahmin_edilen} \nGerçek: {gercek_deger}"
        
    plt.title(baslik, color=renk, fontweight='bold', fontsize=14 )
    plt.axis('off') # Eksen çizgilerini gizle

plt.tight_layout()
plt.show()

print("\nANALİZ: İşte gerçek dünya testi!")
print("Modelimizin başarısı %98 civarında olduğu için büyük ihtimalle yukarıdaki 10 resmin")
print("hepsinin doğru Yeşil renkli olduğunu görüyorsunuz. Bu kodu her çalıştırdığınızda")
print("yepyeni 10 resim gelecek ve yapay zekanın gücünü canlı canlı test etmiş olacaksınız.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("="*75)
print("ZORLU TEST: YAPAY ZEKANIN KAFASINI KARIŞTIRAN RASTGELE 5 HATA")
print("="*75)

# 1. Tüm test setini tahmin ettirelim (Eğer daha önce yapmadıysa)
tum_tahminler_olasilik = model.predict(X_test, verbose=0)
tahmin_edilen_rakamlar = np.argmax(tum_tahminler_olasilik, axis=1)

# 2. Hatalı olanların indekslerini bulalım
hatali_indeksler = np.where(tahmin_edilen_rakamlar != y_test)[0]

print(f"SONUÇ: Toplam {len(X_test)} test resminden {len(hatali_indeksler)} tanesini YANLIŞ bildi.")

# 3. Hatalı indeksler arasından RASTGELE 5 tane seçelim (Değişen Kısım!)
# min(5, len) yapıyoruz ki mucizevi bir şekilde 5'ten az hata yaparsa kod çökmesin :)
kac_tane_gosterilecek = min(10, len(hatali_indeksler))
secilen_hatalar = np.random.choice(hatali_indeksler, size=kac_tane_gosterilecek, replace=False)

print("İşte beyninin yandığı o rastgele rakamlar:\n")

# 4. Seçilen rastgele hatalı resimleri ekrana çizdirelim
plt.figure(figsize=(8, 5))
for i, hata_idx in enumerate(secilen_hatalar):
    plt.subplot(2, 5, i+1)
    
    # Resmi çizdir
    plt.imshow(X_test[hata_idx].reshape(28,28), cmap='gray')
    
    # Gerçek ve Tahmin edilen değeri başlığa yaz
    gercek = y_test[hata_idx]
    tahmin = tahmin_edilen_rakamlar[hata_idx]
    
    plt.title(f"Gerçek: {gercek}\nYapay Zeka: {tahmin}", color='red', fontweight='bold')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

# **Tekrarlayan Sinir Ağları (RNN) ve Zamanın Boyutu**



Geleneksel sinir ağlarında bilgi sadece ileriye doğru akar. RNN'lerde ise bilgi ağın içinde bir **"Döngü" (Loop)** oluşturur. 

Bir RNN hücresi o anki veriyi ($x_t$) alırken, aynı zamanda bir önceki adımdan gelen **"Gizli Durum" (Hidden State - $h_{t-1}$)** bilgisini de alır. Yani kararlarını sadece bugüne bakarak değil, geçmişin tecrübesiyle verir.
$$h_t = \tanh(W_{hh} h_{t-1} + W_{xh} x_t + b_h)$$

### Kısa Vadeli Hafıza Problemi ve LSTM
Standart RNN'ler çok uzun cümlelerde ilk kelimeleri unuturlar. Buna matematikte **"Kaybolan Gradyan" (Vanishing Gradient)** problemi denir. 

Bunu çözmek için RNN'lerin en gelişmiş versiyonu olan **LSTM (Uzun-Kısa Vadeli Hafıza)** icat edilmiştir. 

LSTM hücresinin içinde 3 farklı kapı (Gate) bulunur:
1. **Unutma Kapısı (Forget Gate):** Geçmişteki gereksiz bilgileri (örneğin bitmiş bir cümlenin öznesini) çöpe atar.
2. **Girdi Kapısı (Input Gate):** Yeni gelen bilginin ne kadarının hafızaya yazılacağına karar verir.
3. **Çıktı Kapısı (Output Gate):** Hafızadaki hangi bilgilerin bir sonraki adıma aktarılacağını belirler.

```python
import numpy as np
import matplotlib.pyplot as plt

print("="*75)
print("1. AŞAMA: ZAMAN SERİSİ VERİSİ ÜRETİMİ (SİNÜS DALGASI)")
print("="*75)

# 1. 1000 adımlık bir zaman çizelgesi oluşturalım
zaman = np.arange(0, 100, 0.1)

# 2. Sinüs dalgası üretelim ve üzerine rastgele "gürültü" ekleyelim (Gerçek dünya verisine benzesin)
veri = np.sin(zaman) + np.random.normal(scale=0.1, size=len(zaman))

# 3. RNN'ler için veriyi "Pencerelere" (Lookback) bölmeliyiz.
# Kural: "Geçmiş 20 adıma bak, 21. adımı tahmin et!"
pencere_boyutu = 20
X, y = [], []

for i in range(len(veri) - pencere_boyutu):
    X.append(veri[i : i + pencere_boyutu]) # Geçmiş 20 adım (Özellikler)
    y.append(veri[i + pencere_boyutu])     # Gelecekteki 1 adım (Hedef)

X = np.array(X)
y = np.array(y)

# Keras RNN/LSTM katmanları veriyi 3 boyutlu ister: (Örnek_Sayısı, Zaman_Adımı, Özellik_Sayısı)
X = X.reshape((X.shape[0], X.shape[1], 1))

# Veriyi Eğitim ve Test olarak ikiye bölelim (Geçmişle eğit, Gelecekte test et)
bolme_siniri = int(len(X) * 0.8)
X_train, X_test = X[:bolme_siniri], X[bolme_siniri:]
y_train, y_test = y[:bolme_siniri], y[bolme_siniri:]

# Ürettiğimiz dalganın bir kısmını görelim
plt.figure(figsize=(12, 4))
plt.plot(zaman[:100], veri[:100], label="Gürültülü Dalga", color="blue")
plt.title("Zaman Serisi Verimiz", fontweight="bold")
plt.legend()
plt.show()

print(f"Veri Hazır! Model geçmiş {pencere_boyutu} adıma bakıp, geleceği tahmin etmeye çalışacak.")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

print("\n" + "="*75)
print("2. AŞAMA: YAPAY ZEKAYA HAFIZA EKLEME (LSTM MİMARİSİ)")
print("="*75)

# 1. Modeli İnşa Etme
model = Sequential([
    # Hafıza Katmanı (50 hücreli bir LSTM)
    # input_shape = (Zaman Adımı Sayısı, Özellik Sayısı)
    LSTM(50, activation='relu', input_shape=(pencere_boyutu, 1)),
    
    # Karar Katmanı (Gelecekteki o tek noktayı tahmin etmesi için klasik tek nöronlu Dense)
    Dense(1)
])

# 2. Modeli Derleme
# Regresyon (sayı tahmini) yaptığımız için MSE (Ortalama Karesel Hata) kullanıyoruz.
model.compile(optimizer='adam', loss='mse')

print("Model Özeti:")
model.summary()

print("\nEğitim Başlıyor... (Geçmişin kalıplarını öğreniyor)")
# Modeli eğitiyoruz. Eğitimi hızlı tutmak için 10 epoch (tur) yapıyoruz.
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.1, verbose=1)

print("Eğitim Tamamlandı! Ağımız artık bir hafızaya sahip.")
```

```python
print("\n" + "="*75)
print("3. AŞAMA: GELECEĞİ TAHMİN ETME VE GÖRSELLEŞTİRME")
print("="*75)

# Test verisiyle geleceği tahmin edelim
tahminler = model.predict(X_test)

# Grafik Çizimi (Gerçek Gelecek vs Yapay Zekanın Tahmini)
plt.figure(figsize=(14, 5))

# Sadece test setinin (geleceğin) bir kısmını çizelim ki net anlaşılsın
gosterilecek_uzunluk = 150

plt.plot(y_test[:gosterilecek_uzunluk], label="Gerçek Dalga (Gelecek)", color="blue", linewidth=2)
plt.plot(tahminler[:gosterilecek_uzunluk], label="LSTM Tahmini (Yapay Zeka)", color="red", linestyle="--", linewidth=2)

plt.title("LSTM Ağının Zaman Serisi Tahmini (Geçmişten Geleceği Görmek)", fontsize=14, fontweight="bold")
plt.xlabel("Zaman Adımı")
plt.ylabel("Dalga Değeri")
plt.legend()
plt.grid(True, linestyle=":", alpha=0.7)
plt.show()

print("ANALİZ:")
print("Kırmızı kesik çizgilere dikkat edin! Model sadece rastgele sayılar atmıyor;")
print("öğrendiği hafıza sayesinde dalganın ne zaman yukarı çıkıp ne zaman aşağı ineceğini")
print("neredeyse kusursuz bir şekilde takip ediyor. İşte RNN/LSTM'in gücü budur!")
```

# **Doğal Dil İşleme (NLP): Kelimeleri Matematiğe Çevirmek**

Makinelerin metinleri anlayabilmesi için iki muazzam aşamadan geçmesi gerekir:

### 1. Tokenizasyon (Kelime Dağarcığı Oluşturma)
Bir metindeki her farklı kelimeye benzersiz bir kimlik numarası (ID) verme işlemidir.
* Örnek: "Ben yapay zekayı seviyorum."
* Sözlük: {"ben": 1, "yapay": 2, "zekayı": 3, "seviyorum": 4}
* Çıktı: `[1, 2, 3, 4]`
Ancak bu yeterli değildir. Numaralar sadece bir etikettir, kelimenin "anlamını" içermezler. (2 rakamı 1'den büyük diye "yapay" kelimesi "ben" kelimesinden üstün değildir).

### 2. Kelime Gömme (Word Embeddings - Word2Vec)
[Image of word embeddings vector space showing relationships like King - Man + Woman = Queen]
NLP'nin asıl sihri budur! Kelimeleri uzayda koordinatlara (vektörlere) dönüştürür.
Eğitim sonucunda ağ şunu fark eder: "Kral" ve "Kraliçe" kelimeleri cümle içinde hep benzer yerlerde kullanılıyor. O zaman bunları uzayda birbirine yakın yerlere koyayım.
Hatta öyle bir matematik oluşur ki:
**Vektör("Kral") - Vektör("Erkek") + Vektör("Kadın") = Vektör("Kraliçe")**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

print("="*75)
print("1. AŞAMA: KELİMELERİ SAYILARA ÇEVİRME (TOKENIZATION)")
print("="*75)

# 1. Mini bir Türkçe duygu analizi (Sentiment Analysis) veri seti oluşturalım
cümleler = [
    "Bu film gerçekten harika ve çok güzel",
    "Hayatımda izlediğim en berbat ve kötü film",
    "Oyunculuklar mükemmel, senaryo şahane",
    "Zaman kaybı, iğrenç bir kurgu, nefret ettim",
    "Çok beğendim, kesinlikle tavsiye ederim",
    "Hiç sevmedim, çok sıkıcı ve anlamsız"
]

# Etiketler: 1 (Olumlu), 0 (Olumsuz)
etiketler = np.array([1, 0, 1, 0, 1, 0])

# 2. Tokenizer (Sözlük) Oluşturma
# En çok kullanılan 100 kelimeyi al, bilmediğin kelime gelirse "<OOV>" (Out Of Vocab) yaz.
tokenizer = Tokenizer(num_words=100, oov_token="<OOV>")
tokenizer.fit_on_texts(cümleler)

# Sözlüğün bir kısmına bakalım
kelime_sozlugu = tokenizer.word_index
print(f"Oluşturulan Sözlükteki Toplam Kelime Sayısı: {len(kelime_sozlugu)}")
print("Örnek Numaralandırmalar:", list(kelime_sozlugu.items())[:5])

# 3. Cümleleri Sayı Dizilerine Çevirme
sayisal_diziler = tokenizer.texts_to_sequences(cümleler)
print("\nOrijinal Cümle: ", cümleler[0])
print("Sayısal Hali  : ", sayisal_diziler[0])

# 4. Padding (Hizalama)
# Sinir ağları her cümlenin aynı uzunlukta olmasını ister. Kısa cümlelerin başına 0 ekleriz.
maksimum_uzunluk = 10
hizali_veriler = pad_sequences(sayisal_diziler, maxlen=maksimum_uzunluk, padding='post', truncating='post')

print("\nSinir Ağına Girecek Son Matris (Padding Uygulanmış):")
print(hizali_veriler[0])
print("(Kısa cümlelerin sonuna 0 eklendiğine dikkat edin!)")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

print("\n" + "="*75)
print("🧠 2. AŞAMA: NLP MODELİNİ İNŞA ETME VE EĞİTME (GÜNCELLENDİ)")
print("="*75)

# Modeli Kuruyoruz
model = Sequential([
    Embedding(input_dim=100, output_dim=8, input_length=maksimum_uzunluk),
    LSTM(16, activation='tanh'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("\nYapay Zeka Yoğun Kampa Girdi (500 Tur Eğitim Başladı)...")

# DEĞİŞEN KISIM: 30 tur yetmedi, o yüzden 500 tura (epochs=500) çıkardık!
history = model.fit(hizali_veriler, etiketler, epochs=500, verbose=0)

print("✅ Eğitim Tamamlandı! Modelimiz artık kelimeleri kafasına iyice kazıdı.")
```

```python
print("\n" + "="*75)
print("3. AŞAMA: YAPAY ZEKA İLE CANLI SOHBET (DUYGU ANALİZİ)")
print("="*75)

def duygu_analizi_yap(yeni_cumle):
    # 1. Cümleyi sayılara çevir
    sayisal_hali = tokenizer.texts_to_sequences([yeni_cumle])
    
    # 2. Boyutunu 10 yap (Eksikse 0 ekle, uzunsa kes)
    hizali_hali = pad_sequences(sayisal_hali, maxlen=maksimum_uzunluk, padding='post', truncating='post')
    
    # 3. Modelden tahmin al
    tahmin_olasiligi = model.predict(hizali_hali, verbose=0)[0][0]
    
    # 4. Sonucu Yorumla
    if tahmin_olasiligi > 0.5:
        duygu = "🟢 OLUMLU"
    else:
        duygu = "🔴 OLUMSUZ"
        
    print(f"Cümle: '{yeni_cumle}'")
    print(f"Sonuç: {duygu} (Olumluluk Oranı: %{tahmin_olasiligi*100:.1f})\n")
    print("-" * 50)

# TESTLERİMİZ
print("Modelimizin Test Sonuçları:\n")

# Modelin bildiği kelimelerle test
duygu_analizi_yap("Bu film gerçekten harika")
duygu_analizi_yap("Zaman kaybı ve iğrenç")

# Modelin hiç görmediği bir kombinasyon! (Genellemeyi test ediyoruz)
duygu_analizi_yap("kesinlikle mükemmel tavsiye ederim")

# Siz de istediğiniz cümleyi buraya yazıp test edebilirsiniz (Türkçeye nazaran İngilizce de daha doğru sonuçlar verebilir)!
duygu_analizi_yap("why not")
```

# **Üretken Modeller (Generative Models): Yapay Zekanın Hayal Gücü**

Sınıflandırma modelleri "Bu bir kedi" demeyi öğrenirken, Üretken Modeller **"Kedi çizmeyi"** öğrenirler. Bunu yapmak için verinin *özünü* ve *dağılımını* ezberlerler.

En popüler iki üretken mimari şunlardır:

### 1. Varyasyonel Oto-Kodlayıcılar (VAE - Variational Autoencoders)
Standart bir model veriyi tek bir sayıya sıkıştırırken, VAE veriyi bir **"Olasılık Bulutuna" (Ortalama ve Standart Sapma)** sıkıştırır. Bu sayede o bulutun içinden rastgele bir nokta seçtiğinizde, model o noktayı yepyeni ve mantıklı bir veriye dönüştürür.

### 2. Çekişmeli Üretici Ağlar (GAN - Generative Adversarial Networks)
Polis ve Kalpazan oyunudur. 
* **Üretici (Kalpazan):** Tamamen RASTGELE sayılardan (Gürültü) başlayarak sahte resimler çizer.
* **Ayırt Edici (Polis):** Resmin gerçek mi sahte mi olduğunu anlamaya çalışır.
Üretici o kadar ustalaşır ki, polis artık gerçeği sahteden ayıramaz. Biz de o "Üreticiyi" alır, DALL-E gibi sistemlerin kalbine koyarız!

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.layers import Dense, Reshape

print("="*75)
print("1. AŞAMA: HER ÇALIŞTIRMADA YEPYENİ BİR ŞEKİL UYDURMA")
print("="*75)

# Yapay Zekanın Hayal Gücü Motorunu (Üretici/Generator) Kuruyoruz
ureteci = tf.keras.Sequential([
    Dense(128, activation='relu', input_shape=(32,)), # 32 boyutlu Rastgele Gürültü (Hayal) girecek
    Dense(256, activation='relu'),
    Dense(784, activation='sigmoid'),                 # 784 Piksel (28x28 Resim) çıkacak
    Reshape((28, 28))
])

# (Not: Normalde bu model saatlerce eğitilir. Biz şu an mimariyi anında test etmek 
# için rastgeleliği kullanacağız. Model eğitilmediği için bize "Rakamlara benzeyen uzaylı sembolleri" çizecek!)

# Her çalıştırmada 10 farklı "Rastgele Düşünce" (Gürültü Vektörü) üretiyoruz
rastgele_dusunce = np.random.normal(0, 1, size=(10, 32))

# Üreticiye diyoruz ki: "Bu rastgele sayıları al ve bana ne hayal ettiğini çiz!"
hayali_resimler = ureteci.predict(rastgele_dusunce, verbose=0)

# Ekrana Çizdirme
plt.figure(figsize=(10, 3))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(hayali_resimler[i], cmap='magma') # Her seferinde renkli ve rastgele!
    plt.axis('off')

plt.suptitle("Üretici Ağın Rastgele Hayalleri (Bu hücreyi tekrar tekrar çalıştırın!)", fontweight='bold', fontsize=14)
plt.show()
```

```python
print("\n" + "="*75)
print("2. AŞAMA: SİNİR AĞLARI İLE SOYUT SANAT (CPPN)")
print("="*75)

def yapay_zeka_ressam(boyut=500):
    # 1. Tuvalimizi Hazırlıyoruz (X ve Y koordinatları)
    x = np.linspace(-3, 3, boyut)
    y = np.linspace(-3, 3, boyut)
    X, Y = np.meshgrid(x, y)
    
    # Merkeze olan uzaklığı (Radius) da verelim ki dairesel şekiller çizebilsin
    R = np.sqrt(X**2 + Y**2)
    
    # Girdileri (X, Y, R) yan yana dizip yapay zekaya vereceğiz
    girdiler = np.column_stack([X.ravel(), Y.ravel(), R.ravel()])
    
    # 2. RASTGELE BİR BEYİN (Sinir Ağı) YARATIYORUZ
    # Kuralın gereği: Bu ağırlıklar her çalıştırmada %100 farklı olacak!
    w1 = np.random.randn(3, 32) * 2.0  # İlk gizli katman ağırlıkları
    w2 = np.random.randn(32, 32) * 2.0 # İkinci gizli katman
    w3 = np.random.randn(32, 3) * 2.0  # Çıktı katmanı (R, G, B renkleri için 3 nöron)
    
    # 3. İleri Yayılım (Forward Pass - Fırça Darbeleri)
    # Sinüs (np.sin) ve Hiperbolik Tanjant (np.tanh) aktivasyon fonksiyonları kullanıyoruz
    # Çünkü bu fonksiyonlar organik, dalgalı ve yumuşak kıvrımlar yaratır!
    katman_1 = np.tanh(np.dot(girdiler, w1))
    katman_2 = np.sin(np.dot(katman_1, w2)) 
    
    # Çıktı (Renkler). 0 ile 1 arasına sıkıştırıyoruz ki ekranda gösterilebilsin.
    renkler = 0.5 * (np.tanh(np.dot(katman_2, w3)) + 1.0)
    
    # 1D diziyi tekrar 2D resim formatına (Boyut x Boyut x 3 Renk) çeviriyoruz
    sanat_eseri = renkler.reshape(boyut, boyut, 3)
    return sanat_eseri

# Eseri Üret ve Çizdir
tablo = yapay_zeka_ressam()

plt.figure(figsize=(8, 8))
plt.imshow(tablo)
plt.axis('off')
plt.title("Yapay Zekanın Rastgele Çizdiği Eşsiz Soyut Eser", fontsize=16, fontweight='bold')
plt.show()

print("MUAZZAM DEĞİL Mİ?")
print("Bu resim dünyada SADECE BİR KERE üretildi. Hücreyi bir daha çalıştırırsan,")
print("sinir ağının ağırlıkları (W) değişeceği için tamamen farklı renklerde,")
print("farklı kıvrımlarda yepyeni bir sanat eseri ortaya çıkacak. DALL-E'nin atası!")
```

# **Pekiştirmeli Öğrenme (Reinforcement Learning) ve Q-Learning**



Pekiştirmeli öğrenme, bir köpeğe numara öğretmek gibidir. Köpek (Ajan) doğru bir şey yaparsa ona mama (Ödül) verirsiniz, yanlış bir şey yaparsa kızarsınız (Ceza). Köpek zamanla en çok mamayı nasıl alacağını ezberler.

Bunu matematikte **Q-Learning** algoritması ile yaparız. Yapay zeka beyninde dev bir "Kopya Kağıdı" (Q-Table) tutar. Bu tabloda "Şu konumdayken, şu hareketi yaparsam ne kadar ödül alırım?" sorusunun cevabı puanlanmıştır.

### Bellman Denklemi (Öğrenmenin Matematiği)
Ajan her adım attığında, kopya kağıdındaki o puanı şu efsanevi formülle günceller:
$$Q(s, a) = Q(s, a) + \alpha [R(s, a) + \gamma \max Q(s', a') - Q(s, a)]$$
* **$s$ (State):** Mevcut Durum (Konum)
* **$a$ (Action):** Yapılan Hareket (Sağ, Sol, Yukarı, Aşağı)
* **$R$:** Alınan Anlık Ödül veya Ceza
* **$\alpha$:** Öğrenme Hızı (Yeni bilgiye ne kadar değer verecek?)
* **$\gamma$:** İndirim Faktörü (Gelecekteki ödülleri ne kadar umursayacak?)

```python
import numpy as np
import random
import time
from IPython.display import clear_output

print("="*75)
print("1. AŞAMA: RASTGELE OYUN DÜNYASI (ENVIRONMENT)")
print("="*75)

# 4x4'lük bir oyun alanı (Izgara/Grid)
boyut = 4
durum_sayisi = boyut * boyut
aksiyon_sayisi = 4 # 0: Yukarı, 1: Aşağı, 2: Sol, 3: Sağ

# Her çalıştırmada RASTGELE Konumlar!
# Aynı yere düşmemeleri için rastgele 3 farklı konum seçiyoruz
konumlar = random.sample(range(durum_sayisi), 3)
robot_baslangic = konumlar[0]
hazine_konumu = konumlar[1]
bomba_konumu = konumlar[2]

def haritayi_ciz(robot_konumu):
    """Oyun alanını ekrana çizer"""
    harita = ""
    for i in range(durum_sayisi):
        if i == robot_konumu:
            harita += " 🤖 " # Ajanımız
        elif i == hazine_konumu:
            harita += " 💎 " # Ödül (+10)
        elif i == bomba_konumu:
            harita += " 💣 " # Ceza (-10)
        else:
            harita += " ⬛ " # Boş yol
        
        if (i + 1) % boyut == 0:
            harita += "\n"
    return harita

print(f"Harita Oluşturuldu! Robot: {robot_baslangic}, Hazine: {hazine_konumu}, Bomba: {bomba_konumu}")
print("Oyun Alanımızın Şimdiki Hali:\n")
print(haritayi_ciz(robot_baslangic))
print("Görev: 🤖, 💣'ya basmadan 💎'ı bulmalı!")
```

```python
print("\n" + "="*75)
print("2. AŞAMA: Q-LEARNING İLE EĞİTİM (ÖĞRENME BAŞLIYOR)")
print("="*75)

# Kopya Kağıdını (Q-Table) sıfırlarla oluştur (Satırlar: Konumlar, Sütunlar: Yönler)
q_table = np.zeros((durum_sayisi, aksiyon_sayisi))

# Hiperparametreler
alpha = 0.1      # Öğrenme hızı
gamma = 0.9      # Gelecek planlaması (0-1 arası)
epsilon = 1.0    # Keşif oranı (%100 rastgelelikle başla)
epsilon_azalma = 0.99 # Her oyunda rastgeleliği azalt, öğrendiklerini kullan!

# Hareket Motoru
def hareket_et(konum, aksiyon):
    x = konum % boyut
    y = konum // boyut
    
    if aksiyon == 0 and y > 0: y -= 1       # Yukarı
    elif aksiyon == 1 and y < boyut - 1: y += 1 # Aşağı
    elif aksiyon == 2 and x > 0: x -= 1       # Sol
    elif aksiyon == 3 and x < boyut - 1: x += 1 # Sağ
    return y * boyut + x

# 1000 Tur (Episode) Oyun Oynatıyoruz
for tur in range(1000):
    mevcut_durum = robot_baslangic
    oyun_bitti = False
    
    while not oyun_bitti:
        # 1. Karar Ver (Keşfet veya Bildiğini Oku)
        if random.uniform(0, 1) < epsilon:
            aksiyon = random.randint(0, 3) # Bilmediği için rastgele salla
        else:
            aksiyon = np.argmax(q_table[mevcut_durum]) # Q-Tablosundaki en yüksek puanlı hareketi yap
            
        # 2. Hareketi Gerçekleştir
        yeni_durum = hareket_et(mevcut_durum, aksiyon)
        
        # 3. Ödül/Ceza Sistemi
        if yeni_durum == hazine_konumu:
            odul = 10
            oyun_bitti = True
        elif yeni_durum == bomba_konumu:
            odul = -10
            oyun_bitti = True
        else:
            odul = -0.1 # Her boşa atılan adımda küçük bir ceza ver ki oyalanmasın, hızlı gitsin!
            
        # 4. BELLMAN DENKLEMİ İLE KOPYA KAĞIDINI (Q-TABLE) GÜNCELLE
        eski_deger = q_table[mevcut_durum, aksiyon]
        gelecek_max = np.max(q_table[yeni_durum])
        q_table[mevcut_durum, aksiyon] = eski_deger + alpha * (odul + gamma * gelecek_max - eski_deger)
        
        mevcut_durum = yeni_durum
        
    # Her turdan sonra rastgeleliği biraz daha azalt
    epsilon = max(0.1, epsilon * epsilon_azalma)

print("1000 Tur Eğitim Bitti! Yapay zeka bu haritanın hilesini çözdü.")
print("Q-Table (Kopya Kağıdı) güncellendi. Hadi şimdi canlı izleyelim!")
```

```python
import time
from IPython.display import clear_output

print("\n" + "="*75)
print("3. AŞAMA: YAPAY ZEKA GÖREVDE (CANLI YAYIN)")
print("="*75)

mevcut_durum = robot_baslangic
adim_sayisi = 0

# Animasyon döngüsü
while True:
    clear_output(wait=True)
    print("Oyun Başladı! Yapay zeka öğrendiği yoldan gidiyor...\n")
    print(haritayi_ciz(mevcut_durum))
    
    if mevcut_durum == hazine_konumu:
        print(f"\nMÜKEMMEL! Yapay Zeka Hazineye sadece {adim_sayisi} adımda ulaştı!")
        break
    elif mevcut_durum == bomba_konumu:
        print("\nGÜM! Yapay zeka bombaya bastı (Eğer bunu görüyorsan eğitimin başlarındayız demektir)")
        break
        
    # Rastgelelik yok! Sadece Q-Table'daki en yüksek puanlı (en zekice) hareketi seç
    en_iyi_aksiyon = np.argmax(q_table[mevcut_durum])
    mevcut_durum = hareket_et(mevcut_durum, en_iyi_aksiyon)
    adim_sayisi += 1
    
    time.sleep(0.8) # Adımlar arası 0.8 saniye bekle (Animasyon efekti)
```
